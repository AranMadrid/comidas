```python
import datetime

hora = datetime.datetime.now().hour

if hora >= 7 and hora < 12:
    print("Es la hora del desayuno")
    print("1. Pan con tomate")
    print("2. Tortilla")

    desayuno = input("Elige qué quieres desayunar: ")

    if desayuno == "1":
        print("Has elegido Pan con tomate")

    elif desayuno == "2":
        print("Has elegido Tortilla")

    else:
        print("Opción no válida")


elif hora >= 12 and hora < 17:
    print("Es la hora de la comida")
    print("1. Paella de marisco")
    print("2. Lentejas estofadas")

    comida = input("Elige qué quieres comer: ")

    if comida == "1":
        print("Has seleccionado Paella de marisco")

    elif comida == "2":
        print("Has seleccionado Lentejas estofadas")

    else:
        print("Opción no válida")


elif hora >= 17 and hora < 20:
    print("Es hora de la merienda")
    print("1. Yogur")
    print("2. Bocadillo")

    merienda = input("Elige tu merienda: ")

    if merienda == "1":
        print("Has elegido yogur")

    elif merienda == "2":
        print("Has elegido bocadillo")

    else:
        print("Opción no válida")




```
